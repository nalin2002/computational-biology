## Background

Humanity has a data storage problem. More data has been created in the past few years than in all of preceding history, and that torrent of information may soon outstrip the capacity of traditional storage media. DNA offers a remarkable alternative: a single gram of DNA can theoretically store **215 petabytes** (215 million gigabytes) of data, and it remains stable for thousands of years without requiring energy to maintain [1].

Several research teams have demonstrated proof-of-concept DNA data storage systems, including encoding an entire book [2], all of Wikipedia, and even multimedia files into synthetic DNA [3]. However, the field has not settled on a standard approach to encoding — a search of the literature reveals a wide variety of strategies, each with different tradeoffs.

### How Does It Work?

The basic idea is straightforward: digital data is represented in binary (0s and 1s), and DNA has four nucleotide bases (A, T, G, C). Since four bases can represent more information per character than binary's two digits, DNA is inherently a denser encoding medium. The simplest approach maps pairs of bits to individual bases (e.g., 00→A, 01→T, 10→G, 11→C), converting any digital data — text, images, audio — into a DNA sequence that can be synthesized and stored.

In practice, however, the encoding is more complex. Real DNA synthesis and sequencing introduce constraints and error sources that a good encoding scheme must account for:

- **Homopolymers** — long runs of the same base (e.g., AAAAAAA) are difficult to synthesize and sequence accurately.
- **GC-content balance** — sequences with extreme GC-content (very high or very low) are more error-prone.
- **Error correction** — synthesis and sequencing introduce substitution, insertion, and deletion errors that must be detected and corrected.

Different encoding strategies (lookup tables, Huffman coding, fountain codes, base-3 encoding, and others) each address these constraints in different ways.

### What is a Coding Sequence (CDS)?

A gene's genomic sequence includes **exons**, which are kept in the mRNA, and **introns**, which are spliced out before translation. The **coding sequence (CDS)** is the part of the mRNA that is translated into protein: from the start codon to the stop codon, with the introns already removed. An mRNA record also includes the 5' and 3' untranslated regions (UTRs) on either side of the CDS. Part 3 turns the CDS into sound.

## Part 3: The Song of Your Gene

### Background

DNA encodes a tremendous amount of information — some of which we understand and some that we do not. Researchers and artists have explored many approaches to translating DNA sequences into other representations to look for patterns or simply to experience genetic data in new ways. One of these approaches is **DNA sonification** — converting a DNA sequence into a musical composition.

This problem is creative and exploratory. There is no single "correct" way to map DNA to music, and the goal is for you to design a mapping, implement it, and hear what your gene sounds like.

### The Task

Using the coding sequence you downloaded in Part 2, write a program that takes a DNA sequence and generates a melody with **two layers of sound**:

**Layer 1 — Base notes:** Map each individual nucleotide (A, T, G, C) to a musical note. Each base plays for one beat.

**Layer 2 — Amino acid notes:** Translate the sequence into codons (groups of three bases) and map each resulting amino acid to a musical note. This note plays as a sustained tone underneath all three bases of its codon.

The result is a melody where individual base notes play on top of a slower-moving harmony driven by the amino acid translation:

![DNA Sonification Layers](assets/sonification-layers.png)

Each codon of three bases maps to an amino acid (using the standard genetic code / codon table). The amino acid note sustains for the duration of all three bases in its codon.

### Designing Your Mapping

You are free to choose your own mapping of nucleotides and amino acids to musical notes. Here is one simple example to get you started:

| Nucleotide | Note | Frequency (Hz) |
|------------|------|-----------------|
| A | C4 | 261.63 |
| T | D4 | 293.66 |
| G | E4 | 329.63 |
| C | G4 | 392.00 |

For amino acids, you might assign notes from a lower octave, use a different instrument sound, or map amino acid properties (e.g., hydrophobic vs. polar) to different pitch ranges. Be creative — the mapping is part of your design.

> **Tip:** Stop codons can be handled however you like — silence, a rest, a percussion sound, etc.

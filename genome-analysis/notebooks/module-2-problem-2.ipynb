## Background

A recent study by Wang et al. [1] suggests that Human SARS-CoV-2 has evolved to reduce CG dinucleotide frequency in its open reading frames (ORFs). The authors propose that this reduction in GC-content allows SARS-CoV-2 to reproduce more efficiently in host cells because of reduced RNA stability.

The reasoning is as follows: the stability of RNA stem-loop structures depends on the number of hydrogen bonds formed between bases in the stem. G-C base pairs form **three** hydrogen bonds, while A-U base pairs form only **two**. Therefore, a viral RNA strand with a higher proportion of G and C bases will form more stable (harder to disrupt) stem-loops. A virus with lower GC-content produces less stable stem-loops, meaning less energy is needed to disrupt them during replication — potentially making the virus more efficient at reproducing.

Wang et al. suggest that SARS-CoV-2's lower GC-content compared to other coronaviruses contributes to its higher replication efficiency. Note that replication efficiency does not directly determine fatality — virulence, immune evasion, and host factors all play a role. However, comparing GC-content across coronaviruses with different fatality profiles can still reveal interesting patterns worth examining.

```python
from module_2_assignment.fasta_parser import read_fasta
from module_2_assignment.gc_content import gc_content
from module_2_assignment.identify_orf import ProteinSyn

file_path1= "../data/human_mers-cov.fasta"
file_path2= "../data/human_sars-cov.fasta"
file_path3= "../data/human_sars-cov-2.fasta"

orfs_mers_cov = list()
orfs_sars_cov = list()
orfs_sars_cov_2 = list()

for header, sequence in read_fasta(file_path1):

    protein_syn = ProteinSyn(header, sequence)
    orfs = protein_syn.find_all_orfs()   
    protein_sequences = [orf["protein"] for orf in orfs]
    orfs_mers_cov.extend(protein_sequences) 
    print(f"Header: {header}")
    print(f"No of ORFs: {len(orfs)}")
    print()

for header, sequence in read_fasta(file_path2):

    protein_syn = ProteinSyn(header, sequence)
    orfs = protein_syn.find_all_orfs()   
    protein_sequences = [orf["protein"] for orf in orfs]
    orfs_sars_cov.extend(protein_sequences) 
    print(f"Header: {header}")
    print(f"No of ORFs: {len(orfs)}")
    print()

for header, sequence in read_fasta(file_path3):

    protein_syn = ProteinSyn(header, sequence)
    orfs = protein_syn.find_all_orfs()   
    protein_sequences = [orf["protein"] for orf in orfs]
    orfs_sars_cov_2.extend(protein_sequences) 
    print(f"Header: {header}")
    print(f"No of ORFs: {len(orfs)}")
    print()

gc_avg_mov= sum(gc_content(orf) for orf in orfs_mers_cov)/len(orfs_mers_cov) if orfs else 0.0
gc_avg_sars_cov= sum(gc_content(orf) for orf in orfs_sars_cov)/len(orfs_mers_cov) if orfs else 0.0
gc_avg_sars_cov_2= sum(gc_content(orf) for orf in orfs_sars_cov_2)/len(orfs_mers_cov) if orfs else 0.0

no_of_orfs_mers_cov = len(orfs_mers_cov)
no_of_orfs_sars_cov = len(orfs_sars_cov)
no_of_orfs_sars_cov_2 = len(orfs_sars_cov_2)

print( gc_avg_mov, no_of_orfs_mers_cov)
print( gc_avg_sars_cov, no_of_orfs_sars_cov)
print( gc_avg_sars_cov_2, no_of_orfs_sars_cov_2)
```

Yes, the trend aligns with the hypothesis about replication efficiency.

# Problem 5: Codon position sensitivity
-----------

Using the standard genetic code, determine which codon position — first, second, or third — is most sensitive to single-nucleotide substitutions.

For this question, treat each sense codon equally. For each codon position, consider all three possible single-nucleotide substitutions at that position and determine whether the substitution changes the encoded amino acid. A substitution that produces a stop codon counts as changing the amino acid.

## Calculate
1.	The fraction of substitutions at each codon position that change the amino acid
2.	Which position is most sensitive
3.	How many times more likely an amino-acid change is at the most sensitive position than at the least sensitive position

Explain your reasoning and include a table and/or summary of your results. The correct answer will require computation, not intuition.

```python
# standard codon table
codon_table = {
    "TTT":"F", "TTC":"F", "TTA":"L", "TTG":"L",
    "TCT":"S", "TCC":"S", "TCA":"S", "TCG":"S",
    "TAT":"Y", "TAC":"Y", "TAA":"*", "TAG":"*",
    "TGT":"C", "TGC":"C", "TGA":"*", "TGG":"W",

    "CTT":"L", "CTC":"L", "CTA":"L", "CTG":"L",
    "CCT":"P", "CCC":"P", "CCA":"P", "CCG":"P",
    "CAT":"H", "CAC":"H", "CAA":"Q", "CAG":"Q",
    "CGT":"R", "CGC":"R", "CGA":"R", "CGG":"R",

    "ATT":"I", "ATC":"I", "ATA":"I", "ATG":"M",
    "ACT":"T", "ACC":"T", "ACA":"T", "ACG":"T",
    "AAT":"N", "AAC":"N", "AAA":"K", "AAG":"K",
    "AGT":"S", "AGC":"S", "AGA":"R", "AGG":"R",

    "GTT":"V", "GTC":"V", "GTA":"V", "GTG":"V",
    "GCT":"A", "GCC":"A", "GCA":"A", "GCG":"A",
    "GAT":"D", "GAC":"D", "GAA":"E", "GAG":"E",
    "GGT":"G", "GGC":"G", "GGA":"G", "GGG":"G"
}

bases = ["A", "C", "G", "T"]

# Only consider the 61 sense codons as starting codons and ignoring stop codons
sense_codons = [
    codon for codon, amino_acid in codon_table.items()
    if amino_acid != "*"
]

# Store number of amino-acid changes and total substitutions for positions 1, 2, and 3
changes = [0, 0, 0]
total = [0, 0, 0]

for codon in sense_codons:

    original_amino_acid = codon_table[codon]

    # Testing each of the three positions
    for position in range(3):

        # Replacing the nucleotide with each possible base
        for base in bases:

            # skipping the original base
            if base == codon[position]:
                continue

            # Creating the mutated codon
            mutated_codon = list(codon)
            mutated_codon[position] = base
            mutated_codon = "".join(mutated_codon)

            mutated_amino_acid = codon_table[mutated_codon]

            total[position] += 1

            # counting the amino acid changes
            if mutated_amino_acid != original_amino_acid:
                changes[position] += 1


# Calculate fraction of substitutions that change the amino acid
fractions = [
    changes[i] / total[i]
    for i in range(3)
]

print("Position | Changes | Total | Fraction | Percentage")
print("-" * 55)

for i in range(3):
    print(
        f"{i + 1:^8} | "
        f"{changes[i]:^7} | "
        f"{total[i]:^5} | "
        f"{fractions[i]:.4f}   | "
        f"{fractions[i] * 100:.2f}%"
    )

# Find most and least sensitive positions
most_sensitive = fractions.index(max(fractions))
least_sensitive = fractions.index(min(fractions))

# Calculating how many times more sensitive
ratio = (
    fractions[most_sensitive]
    / fractions[least_sensitive]
)

print()
print(f"Most sensitive position: {most_sensitive + 1}")
print(f"Least sensitive position: {least_sensitive + 1}")
print(f"Most/least sensitivity ratio: {ratio:.2f}x")
```

Reasoning:

I considered all 61 sense codons in the standard genetic code. For each codon, I independently mutated the first, second, and third positions. At each position there are three possible substitutions because the original nucleotide can be replaced by any of the other three bases. I evaluated 183 ( 61*3 ) possible substitutions at each codon position.

For each substitution, I compared the amino acid encoded by the mutated codon with the amino acid encoded by the original codon. If they were different, I counted it as an amino-acid change. Mutations producing a stop codon were also counted as changes.

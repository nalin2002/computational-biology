FASTA format is a file format used to exchange information between genetic sequence databases.  A sequence in FASTA format begins with a single-line description, followed by lines of sequence data.  The description line is distinguished from the sequence data by a greater-than ("`>`") symbol in the first column.   

The word following the "`>`" symbol is the identifier of the sequence, and the rest of the line is the description (both are optional). There should be no space between the ">" and the first letter of the identifier. It is recommended that all lines of text be shorter than 80 characters. No blank lines are allowed in the middle of the FASTA input.  The sequence ends if another line starting with a "`>`" appears; this indicates the start of another sequence. 

Normally, identifiers are simply protein accession, name or Entrez gi's (e.g., `Q5I7T1`, `AG10B_HUMAN`, `129295`), but a bar-separated NCBI sequence identifier (e.g., `gi|129295`) will also be accepted. Any arbitrary user-specified sequence identifier can also be used (e.g., CLONE00073452) but you are advised to use sufficiently long unique words in such case.

Sequences are expected to be represented in the standard IUB/IUPAC amino acid and nucleic acid codes, with these exceptions:
- Lower-case letters are accepted and are mapped into upper-case;
- A single hyphen or dash can be used to represent a gap of indeterminate length;
- In amino acid sequences, U and * are acceptable letters (see below).
- Any numerical digits in the query sequence should either be removed or replaced by appropriate letter codes (e.g., N for unknown nucleic acid residue or X for unknown amino acid residue).

The standard IUB/IUPAC nucleic acid codes are:

        A --> adenine             M --> A C (amino)
        C --> cytosine            S --> G C (strong)
        G --> guanine             W --> A T (weak)
        T --> thymine             B --> G T C
        U --> uracil              D --> G A T
        R --> G A (purine)        H --> A C T
        Y --> T C (pyrimidine)    V --> G C A
        K --> G T (keto)          N --> A G C T (any)
                                  -  gap of indeterminate length

There is no standard file extension for a text file containing FASTA formatted sequences. Some commonly used extensions are `.fasta, .fas, and .fa`.

Write a FASTA file parser that accomplishes the following tasks:
- Reads a FASTA formatted file from disk
- Validates its formatting and composition (i.e. does not contain any incorrect characters). The parser should be able to parse DNA, RNA and protein sequences.  If a record is invalid, skip it, print a warning that identifies the record, and continue with the rest of the file.
- Store and return the description and sequence in a tuple, e.g. `(description, sequence)`. For this problem, the description is all the text on the line that starts with `>`, after the `>` itself: the identifier and the rest of the line together. For the example above, it is `gi|568815587:c5227071-5225466 Homo sapiens chromosome 11, GRCh38 Primary Assembly`
- The parser should be capable of processing multiple sequences in a single file.  In some scenarios, a file could have hundreds of thousands of sequences and may be impractical (or impossible) to read them all into memory.  You must employ an approach to account for this. Here are some approaches to handling this type of scenario in Python: [Yield and Generators Explained](http://www.jeffknupp.com/blog/2013/04/07/improve-your-python-yield-and-generators-explained/) or [Python Yields are Fun!](http://pydanny.com/python-yields-are-fun.html)

## Rules
Your parser follows these rules, which are more lenient than the format description above, as most real parsers are:
- Remove whitespace from the end of every line, including the `\r` of Windows line endings, before checking it
- Ignore blank lines anywhere in the file
- Lines before the first `>` are an error: print a warning and skip them
- A header with no sequence lines is an invalid record: print a warning and skip it, as for any other invalid record

Use your parser to read in the following files:

Print out the description and sequences in the following format: 
```
- file_name | description | sequence_length | sequence
- file_name | description | sequence_length | sequence  
- file_name | description | sequence_length | sequence  
- file_name | description | sequence_length | sequence  
```

````python
import warnings

# Considering valid characters for each type sa per the table above
VALID_DNA = set("ACGTMSWBDRHVYKN-")
VALID_RNA = set("ACGUMSWBDRHVYKN-")
VALID_PROTEIN = set("ABCDEFGHIKLMNPQRSTUVWYZX*-")

# Checking whether a sequence contains valid biological characters
def check_valid(sequence):

    # if sequence is empty, its invalid
    if not sequence:
        return False

    sequence = sequence.upper()

    # a sequence is valid, when all of its characters are present in atleast one of the 3 types above
    if all(char in VALID_DNA for char in sequence):
        return True

    if all(char in VALID_RNA for char in sequence):
        return True

    if all(char in VALID_PROTEIN for char in sequence):
        return True

    return False

# Classifying the type of sequence 
def classify(sequence):

    if not sequence:
        return "Unknown"

    sequence = sequence.upper()

    if all(char in VALID_DNA for char in sequence):
        return "DNA"

    elif all(char in VALID_RNA for char in sequence):
        return "RNA"

    elif all(char in VALID_PROTEIN for char in sequence):
        return "Protein"

    else:
        return "Unknown"

def read_fasta(file_path):

    description = None
    sequence = ""

    with open(file_path, "r") as f:

        for line_number, line in enumerate(f, start=1):

            # removing whitespace from end of line
            line = line.rstrip()

            # Ignoring blank lines
            if not line:
                continue

            # Reading the header line
            if line.startswith(">"):

                # If we already have a record, we process it first before starting a new one
                if description is not None:

                    if not check_valid(sequence):
                        warnings.warn(
                            f"Invalid sequence found for "
                            f"'{description}'. Skipping it."
                        )
                    else:
                        seq_type = classify(sequence)
                        if seq_type == "Unknown":
                            warnings.warn(
                                f"Unknown sequence type found for "
                                f"'{description}'. Skipping it."
                            )
                        else:
                            yield (description, sequence)

                # Starting a new record , everything after ">" is description
                description = line[1:]

                # reseting sequence for new record
                sequence = ""

            # Reading the sequence lines
            else:

                # Sequence data before first header is invalid
                if description is None:
                    warnings.warn(
                        f"Line {line_number}: sequence data found "
                        f"before first FASTA header. Skipping line."
                    )
                    continue

                # Adding line to current sequence
                sequence += line.upper()

        # Processing the final fasta record
        if description is not None:

            if not check_valid(sequence):
                warnings.warn(
                    f"Invalid sequence found for "
                    f"'{description}'. Skipping it."
                )
            else:
                seq_type = classify(sequence)
                if seq_type == "Unknown":
                    warnings.warn(
                        f"Unknown sequence type found for "
                        f"'{description}'. Skipping it."
                    )
                else:
                    yield (description, sequence)


# Input file paths
file1_path = "../data/multiple_DNA.fasta"
file2_path = "../data/multiple_protein.fasta"
file3_path = "../data/spike-protein-dna.fasta"
file4_path = "../data/spike-protein-protein.fasta"


file_paths = [
    file1_path,
    file2_path,
    file3_path,
    file4_path
]


for file_path in file_paths:
    print(f"\nReading: {file_path}")
    for description, sequence in read_fasta(file_path):
        print(
            f"{file_path} | "
            f"{description} | "
            f"{len(sequence)} | "
            f"{sequence[:30]}..."
        )

# Classification Tests
# DNA
assert classify("ATGCGT") == "DNA"

# RNA
assert classify("AUGCGU") == "RNA"

# Protein
assert classify("MKWVTF") == "Protein"

# Empty sequence
assert classify("") == "Unknown"

# Ambiguous sequence, priority is DNA -> RNA -> Protein and hence this is classified first as a DNA
assert classify("ACGACG") == "DNA"

print("All tests passed.")
````

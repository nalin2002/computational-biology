> Place all files for this problem in a directory called `problem_2`.

## Overview

Interactive sessions are useful for development and debugging, but real workloads on an HPC cluster are submitted as **batch jobs**. A batch job is defined in a shell script (an `.sbatch` file) that tells the SLURM scheduler what resources you need and what commands to run. In this problem, you will write and submit your first batch job, learn to monitor its status, and inspect its output.


## Part A: Write a Batch Script

Create a file called `hello_batch.sbatch` with the following structure:

```bash
#!/bin/bash
#SBATCH --job-name=hello_batch
#SBATCH --account=mpcs56430
#SBATCH --output=hello_batch.out
#SBATCH --error=hello_batch.err
#SBATCH --partition=caslake
#SBATCH --nodes=1
#SBATCH --ntasks=1
#SBATCH --mem=2G
#SBATCH --time=00:10:00

# Your commands go below this line
```

Add commands to the script that perform the following:

1. Print the job ID (`$SLURM_JOB_ID`)
2. Print the node the job is running on (`$SLURM_NODELIST`)
3. Print the number of tasks (`$SLURM_NTASKS`)
4. Print the current date and time
5. Add a `sleep 60` command so the job runs long enough for you to observe it in the queue
6. Load Python and run your `hello_cluster.py` from Problem 1 (copy it into this directory)
7. Print the date and time again after the script completes


## Part B: Submit and Monitor

1. Submit the job:

```bash
sbatch hello_batch.sbatch
```

Record the job ID that SLURM returns.

2. Immediately after submitting, check the status of your job:

```bash
squeue --user=<your_cnetid>
```

3. In a file called `answers.md`, answer the following:

   - What does the `ST` (state) column show while the job is waiting? While it is running?
   - What does the `NODELIST(REASON)` column show while the job is pending?
   - How long did the job spend in the queue before it started running? (Hint: compare submit time and start time)


## Part C: Inspect the Output

After the job completes:

1. Examine the contents of `hello_batch.out` and `hello_batch.err`.

2. Run `sacct` to get details about your completed job:

```bash
sacct --jobs=<your_job_id> --format=JobID,JobName,Partition,State,ExitCode,Elapsed,MaxRSS
```

3. In `answers.md`, record the following from the `sacct` output:

   - What was the **State** of the job?
   - What was the **ExitCode**?
   - How much wall time (**Elapsed**) did the job use?
   - How much memory (**MaxRSS**) did the job actually consume?
   - How does the actual memory usage compare to the 2G you requested?


## Part D: Experiment with SBATCH Directives

Modify your batch script (or create a second one called `hello_batch_v2.sbatch`) to explore how different directives affect behavior:

1. Change the `--output` directive to use SLURM filename patterns:

```bash
#SBATCH --output=hello_%j.out
```

   What does `%j` resolve to in the output filename?

2. Submit the modified job and confirm the output filename matches expectations.

Record your observations in `answers.md`.

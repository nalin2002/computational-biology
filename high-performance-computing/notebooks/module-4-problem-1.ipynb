## Overview

Before running parallel jobs on a high-performance computing (HPC) cluster, you need to understand how to connect, navigate the environment, and run commands interactively. In this problem, you will log in to the RCC Midway cluster, request an interactive session, explore the filesystem layout, and run a simple Python script — all before writing a single batch script.


## Part A: Connect and Explore

1. Log in to the RCC Midway cluster using SSH:

```bash
ssh <your_cnetid>@midway3.rcc.uchicago.edu
```

2. Once logged in, answer the following questions in a file called `answers.md`:

   - What is the full path of your **home directory**?
   - What is the full path of your **scratch directory**? (Hint: check `/scratch/midway3/<your_cnetid>`)
   - Run `quota` or `rcchelp quota` — what are the storage limits on your home directory versus scratch?
   - Run `module avail python` — what versions of Python are available on the cluster?

3. Create a personal directory under `/project/` using your CNetID:

```bash
mkdir -p /project/<your_cnetid>
```

This will serve as your working project space for assignments throughout the course. Verify the directory was created and record its full path in `answers.md`.

4. Run `squeue` to view all jobs currently in the cluster queue:

```bash
squeue | wc -l
```

How many jobs are currently in the queue? Record the count and the time you checked in `answers.md`. Note that this includes jobs in all states — pending, running, and otherwise.

5. How many **nodes** are available on the Midway cluster? Record the total count in `answers.md`.

   > Hint: The `sinfo` command provides information about nodes and partitions. Check `sinfo --help` or `man sinfo` to find flags that can help you list individual nodes and produce a clean count.

6. Copy your `answers.md` file from the cluster back to your local computer so you can submit it. There are several ways to transfer files from a remote server to your local machine. Research and try at least **two** of the following methods:

   - **`scp`** — Secure copy over SSH
   - **`rsync`** — Synchronization tool that works over SSH
   - **`sftp`** — Interactive secure file transfer session
   - **VS Code Remote SSH** — Edit and save remote files directly from your editor
   - **`sshfs`** — Mount the remote filesystem locally so it appears as a folder on your machine

   In `answers.md`, briefly describe which two methods you used and note any pros or cons you observed.


## Part B: Request an Interactive Session

On a shared cluster, the **login nodes** are not meant for running computational work. Instead, you must request compute resources. The `sinteractive` command is the simplest way to get an interactive shell on a compute node.

1. Request an interactive session with the following resources:

```bash
sinteractive --account mpcs56430 --partition=caslake --nodes=1 --ntasks=1 --mem=4G --time=00:30:00
```

2. Once your session starts, confirm you are on a **compute node** (not a login node):

```bash
hostname
```

3. Record the hostname in `answers.md`. How does it differ from the login node hostname?

4. While in the interactive session, check the available CPUs and memory:

```bash
nproc
free -h
```

Record these values in `answers.md`. Do they match what you requested with your `sinteractive` command? You will likely notice the numbers are much larger than expected — that is because `nproc` and `free -h` report the **total hardware** on the node, not what SLURM allocated to your job.

When SLURM starts a job, it sets a number of **environment variables** that describe the resources allocated to that specific job. Some useful ones include:

| Variable                | Description                                  |
| ----------------------- | -------------------------------------------- |
| `$SLURM_JOB_ID`        | The unique ID assigned to your job           |
| `$SLURM_NTASKS`        | Number of tasks allocated                    |
| `$SLURM_MEM_PER_NODE`  | Memory allocated per node (in MB)            |
| `$SLURM_NODELIST`      | Name(s) of the node(s) assigned to your job  |
| `$SLURM_SUBMIT_DIR`    | The directory from which the job was submitted|

Try printing a few of these in your interactive session:

```bash
echo $SLURM_JOB_ID
echo $SLURM_NTASKS
echo $SLURM_MEM_PER_NODE
```

In `answers.md`, compare the values from `nproc`/`free -h` with the SLURM variables. Which set of values reflects what your job was actually given?


## Part C: Run a Script Interactively

1. Check that Python is available by running `python3 --version`. A system Python should already be accessible on the compute node. If you need additional packages beyond the standard library later in the course, you can load the Anaconda distribution with:

```bash
module load python/anaconda-2024.02
```

2. Create a Python script called `hello_cluster.py` that does the following:

   - Prints a greeting message
   - Prints the current hostname (use `import socket; socket.gethostname()`)
   - Prints the current working directory
   - Prints the number of available CPUs (use `import os; os.cpu_count()`)
   - Prints the current date and time

3. Run the script inside your interactive session:

```bash
python hello_cluster.py
```

4. Save the output in a file called `interactive_output.txt`.

5. Exit the interactive session:

```bash
exit
```

## Overview

Understanding the filesystem layout on an HPC cluster is critical for running jobs efficiently and avoiding common pitfalls. The RCC Midway cluster has multiple storage locations, each with different performance characteristics, capacity limits, and retention policies. In this problem, you will learn to navigate these filesystems, transfer data to and from the cluster, and write jobs that properly stage data for computation.


## Part A: Filesystem Layout

Log in to the cluster and investigate the storage locations available to you. In a file called `answers.md`, document the following:

1. **Home directory** (`~/` or `/home/<cnetid>`)
   - What is the storage quota?
   - Is this space backed up?
   - Is this space suitable for running large computational jobs? Why or why not?

2. **Scratch space** (`/scratch/midway3/<cnetid>`)
   - What is the storage quota?
   - What is the **purge policy** — how long can files remain unused before they are deleted?
   - Why does scratch exist as a separate space from home?

3. **Project space** (if your course has one, e.g., `/project/mpcs56420` or similar)
   - How does project space differ from home and scratch?

> Hint: Use `rcchelp quota` and consult the [RCC User Guide](https://rcc-uchicago.github.io/user-guide/) if needed.


## Part B: Transfer Files to the Cluster

From your **local machine**, practice transferring your Module 3 assignment to the cluster using `scp` and `rsync`.

1. Use `scp` to copy a single file from your Module 3 assignment directory to your scratch directory on the cluster:

```bash
scp path/to/module-3/<a_file> <your_cnetid>@midway3.rcc.uchicago.edu:/scratch/midway3/<your_cnetid>/
```

2. Now use `rsync` to transfer your entire Module 3 assignment directory:

```bash
rsync -avz path/to/module-3/ <your_cnetid>@midway3.rcc.uchicago.edu:/scratch/midway3/<your_cnetid>/module-3/
```

3. Log in to the cluster and verify the files arrived in your scratch directory.

4. In `answers.md`, describe the difference between `scp` and `rsync`. When would you prefer one over the other?


## Part C: Data Staging in a Batch Job

A common pattern in HPC is to **stage** data — copy input files from a persistent location to a fast local or scratch space before computation, and copy results back when finished.

In this part, you will run your **GC-content analysis script** from Problem 2 of Assignment 2 (Coronavirus GC-Content Analysis) as a batch job using the data staging pattern.

1. Ensure your GC-content script and the coronavirus FASTA files (`human_mers_cov.fasta`, `human_sars-cov.fasta`, `human_sars_cov-2.fasta`) are in your Module 3 directory on the cluster (which you transferred in Part B).

2. Write a batch script called `stage_job.sbatch` that does the following:

   - Creates a temporary working directory in scratch using the job ID:
     ```bash
     WORK_DIR=/scratch/midway3/<your_cnetid>/job_${SLURM_JOB_ID}
     mkdir -p $WORK_DIR
     ```
   - Copies your GC-content script and the coronavirus FASTA files into `$WORK_DIR`
   - Changes into `$WORK_DIR`
   - Runs your GC-content script against the three coronavirus genomes
   - Copies the output back to a `results/` directory in your home directory
   - Cleans up the temporary working directory

3. Submit the job and verify that:
   - The results appear in `~/results/`
   - The temporary directory in scratch has been removed

4. In `answers.md`, explain why this stage-in / stage-out pattern is considered good practice on shared clusters.

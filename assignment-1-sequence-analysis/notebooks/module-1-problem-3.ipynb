# Problem 3: Base composition of the SARS-CoV-2 genome
-----------

Plot the distribution of the `A`,`G`,`C`,`T` bases for the SARS-CoV2 genome (`data/sars-cov2-genome.fasta`) using matplotlib.  Show the plot _inline_ in the notebook.

> Note that the problem is to plot the data, not necessarily write a program. Use any tools at your disposal to accomplish this.

```python
%matplotlib inline

import matplotlib.pyplot as plt

def read_fasta(file_path):
    count_bases = {
        'A': 0,
        'G': 0,
        'C': 0,
        'T': 0
    }
    # finding the distribution by counting the base characters in all sequences, ignoring the headers
    with open(file_path, 'r') as f:
        for line in f:
            line = line.strip()

            if not line: # Skipping empty lines
                continue

            if line.startswith(">"): # Skipping headers
                continue

            # Count bases in this sequence line
            for base in line.upper():
                if base in count_bases:
                    count_bases[base] += 1

    return count_bases


base_counts = read_fasta("../data/sars-cov2-genome.fasta")

print(base_counts)

plt.bar(base_counts.keys(), base_counts.values())
plt.xlabel("Base")
plt.ylabel("Count")
plt.title("SARS-CoV-2 Genome Base Distribution")
plt.show()
```

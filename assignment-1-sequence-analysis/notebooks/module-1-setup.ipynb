# 4. Set Up Your Development Environment
---
* **GitHub.** Sign up for a [GitHub](https://github.com/) account to use for this course, or use one you already have.
    * Course repositories are created in the course's GitHub organization, `uchicago-scientific`, not under your account. GitHub emails you an invitation to each one.
* **Everything else.** Follow [Your working environment](https://uchicago-scientific.github.io/mpcs56430-handbook/setup/) in the course handbook. It covers WSL for Windows users, [uv](https://docs.astral.sh/uv/) (which installs Python for you), git and the GitHub CLI, and VS Code with its Python and Jupyter extensions.
* Check the setup by running `uv --version`, `git --version` and `gh auth status` in a new terminal.
* **(Optional)** Download and install [GitHub Desktop](https://desktop.github.com/).

> The instructors use VS Code throughout the course. You can use a different editor, but you will receive limited support.

# 6. UNIX Refresher
---
Utilizing the command line is an essential skill in scientific computing. After this week, the expectation will be that you can work comfortably from the command line and be able to construct and execute bash scripts. Start with [The command line](https://uchicago-scientific.github.io/mpcs56430-handbook/command-line/) in the course handbook. For more practice:

* [Tutorial - Linux Basics — UChicago CS Student Resource Guide](https://uchicago-cs.github.io/student-resource-guide/tutorials/linux-basics.html)

* [Command Line Crash Course](https://learnpythonthehardway.org/python3/appendixa.html): 15 exercises for common command line techniques.
    > If you are using macOS, then follow the commands for Linux.

# 8. Familiarize Yourself with a Text-based Editor
---
We will be utilizing remote computing resources that will constrain us to working with classic (i.e., text-based) editors. While there is no general consensus on which is the best, there are three options that are universal and available across all platforms. Please familiarize yourself with one of them. You really just need to be comfortable with opening, editing, and saving files. [Editing files in a terminal](https://uchicago-scientific.github.io/mpcs56430-handbook/text-editors/) in the course handbook covers nano and Vim.
